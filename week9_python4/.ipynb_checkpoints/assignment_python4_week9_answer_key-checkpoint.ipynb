## Setup — choose the course FASTA file

````python
import re
import sys

# Start with the manakin data. Later, change this to the pine dataset.
InputFile = "Manacus_vitellinus.gene.cds.fa"
print(InputFile)
# InputFile = "taeda_unigenes.fasta"
````

## Task 1 — Count restriction enzyme recognition sites

#### A. Count the sequences
Count the total number of FASTA records in the file. Remember that each identifier line beginning with `>` represents a sequence. Initialize your counter **before** the loop and increment it as you read the file.

#### B. Count restriction sites
Count the number of matches to each motif across all sequences in the file: **EcoRI** (`GAATTC`), **MseI** (`TTAA`), and **HindIII** (`AAGCTT`). Use `len(re.findall(...))` to count matches within one sequence and add these counts to running totals. Count nonoverlapping matches on the sequence as written.

#### C. Summarize your results
Print a clearly labeled summary including the input filename, number of sequences, and total matches for each restriction site. You may also write the summary to a text file.

````python
# Task 1, A–C: deliberately similar to examples in the primers.
IN = open(InputFile, 'r')

SeqCount = 0
EcoRI = 0
MseI = 0
HindIII = 0

for Line in IN:
    Fasta = Line.strip("\n")    # identifier line
    Seq = IN.readline()           # read the next line: DNA sequence
    Seq = Seq.strip("\n")

    SeqCount += 1
    EcoRI += len(re.findall('GAATTC', Seq))
    MseI += len(re.findall('TTAA', Seq))
    HindIII += len(re.findall('AAGCTT', Seq))

IN.close()

print("Input file:", InputFile)
print("Number of sequences:", SeqCount)
print("EcoRI (GAATTC):", EcoRI)
print("MseI (TTAA):", MseI)
print("HindIII (AAGCTT):", HindIII)
````

## Task 2 — Find microsatellite repeats

Microsatellites (SSRs) are short sequence motifs repeated consecutively. Count **distinct repeat regions (loci)**, not individual repeat units.

#### A. Count AT repeats
Count the number of distinct regions containing at least four consecutive `AT` repeats.

#### B. Count ATC repeats
Count the number of distinct regions containing at least four consecutive `ATC` repeats.

#### C. Count GA repeats
Count the number of distinct regions containing at least four consecutive `GA` repeats.

#### D. Export sequences containing SSRs
Create a new FASTA file containing every sequence with at least one of these three SSR types. Write its original identifier and sequence, **once per record**, even if it has multiple SSRs. Print the number of sequences exported.

### A–C. Count microsatellite loci across the entire file

````python
IN = open(InputFile, 'r')

ATtotal = 0
ATCtotal = 0
GAtotal = 0
SeqCount = 0

for Line in IN:
    Fasta = Line.strip("\n")
    Seq = IN.readline()
    Seq = Seq.strip("\n")

    SeqCount += 1
    ATtotal += len(re.findall(r'(?:AT){4,}', Seq))
    ATCtotal += len(re.findall(r'(?:ATC){3,}', Seq))
    GAtotal += len(re.findall(r'(?:GA){4,}', Seq))

IN.close()

print("Input file:", InputFile)
print("Number of sequences:", SeqCount)
print("AT microsatellite loci:", ATtotal)
print("ATC microsatellite loci:", ATCtotal)
print("GA microsatellite loci:", GAtotal)
````

### D. Write sequences with SSRs to a new FASTA file

````python
IN = open(InputFile, 'r')
OutFile = InputFile + '.SSRs.fa'
OUT = open(OutFile, 'w')

Exported = 0

for Line in IN:
    Fasta = Line.strip("\n")
    Seq = IN.readline()
    Seq = Seq.strip("\n")

    # One conditional for all three SSR types: a record is written only once.
    if (re.search(r'(?:AT){4,}', Seq) or
        re.search(r'(?:ATC){4,}', Seq) or
        re.search(r'(?:GA){4,}', Seq)):
        OUT.write(Fasta + "\n")
        OUT.write(Seq + "\n")
        Exported += 1

IN.close()
OUT.close()

print("Input file:", InputFile)
print("Sequences exported:", Exported)
print("Output FASTA:", OutFile)
````

## Full solutions for **both** course datasets

The next two sections run the complete Task 1 and Task 2 analysis separately for each dataset. Each uses explicit file handles, running totals, and a loop over two-line FASTA records. They also write a text summary and a FASTA file containing SSR-bearing sequences. No functions or more advanced data structures are required.

### Manakin — *Manacus vitellinus*: Tasks 1A–C and 2A–D

````python
InputFile = "Manacus_vitellinus.gene.cds.fa"
IN = open(InputFile, 'r')
OUT = open(InputFile + '.SSRs.fa', 'w')

SeqCount = 0
EcoRI = 0
MseI = 0
HindIII = 0
ATtotal = 0
ATCtotal = 0
GAtotal = 0
Exported = 0

for Line in IN:
    Fasta = Line.strip("\n")
    Seq = IN.readline()
    Seq = Seq.strip("\n")

    SeqCount += 1
    EcoRI += len(re.findall('GAATTC', Seq))
    MseI += len(re.findall('TTAA', Seq))
    HindIII += len(re.findall('AAGCTT', Seq))

    ATtotal += len(re.findall(r'(?:AT){4,}', Seq))
    ATCtotal += len(re.findall(r'(?:ATC){4,}', Seq))
    GAtotal += len(re.findall(r'(?:GA){4,}', Seq))

    if (re.search(r'(?:AT){4,}', Seq) or
        re.search(r'(?:ATC){4,}', Seq) or
        re.search(r'(?:GA){4,}', Seq)):
        OUT.write(Fasta + "\n")
        OUT.write(Seq + "\n")
        Exported += 1

IN.close()
OUT.close()

Summary = (f"Input file: {InputFile}\n"
           f"Number of sequences: {SeqCount}\n"
           f"EcoRI (GAATTC): {EcoRI}\n"
           f"MseI (TTAA): {MseI}\n"
           f"HindIII (AAGCTT): {HindIII}\n"
           f"AT SSR loci: {ATtotal}\n"
           f"ATC SSR loci: {ATCtotal}\n"
           f"GA SSR loci: {GAtotal}\n"
           f"Sequences exported: {Exported}\n")

print(Summary)
REPORT = open(InputFile + '.summary.txt', 'w')
REPORT.write(Summary)
REPORT.close()
````

### Loblolly pine — *Pinus taeda*: Tasks 1A–C and 2A–D

````python
InputFile = "taeda_unigenes.fasta"
IN = open(InputFile, 'r')
OUT = open(InputFile + '.SSRs.fa', 'w')

SeqCount = 0
EcoRI = 0
MseI = 0
HindIII = 0
ATtotal = 0
ATCtotal = 0
GAtotal = 0
Exported = 0

for Line in IN:
    Fasta = Line.strip("\n")
    Seq = IN.readline()
    Seq = Seq.strip("\n")

    SeqCount += 1
    EcoRI += len(re.findall('GAATTC', Seq))
    MseI += len(re.findall('TTAA', Seq))
    HindIII += len(re.findall('AAGCTT', Seq))

    ATtotal += len(re.findall(r'(?:AT){4,}', Seq))
    ATCtotal += len(re.findall(r'(?:ATC){4,}', Seq))
    GAtotal += len(re.findall(r'(?:GA){4,}', Seq))

    if (re.search(r'(?:AT){4,}', Seq) or
        re.search(r'(?:ATC){4,}', Seq) or
        re.search(r'(?:GA){4,}', Seq)):
        OUT.write(Fasta + "\n")
        OUT.write(Seq + "\n")
        Exported += 1

IN.close()
OUT.close()

Summary = (f"Input file: {InputFile}\n"
           f"Number of sequences: {SeqCount}\n"
           f"EcoRI (GAATTC): {EcoRI}\n"
           f"MseI (TTAA): {MseI}\n"
           f"HindIII (AAGCTT): {HindIII}\n"
           f"AT SSR loci: {ATtotal}\n"
           f"ATC SSR loci: {ATCtotal}\n"
           f"GA SSR loci: {GAtotal}\n"
           f"Sequences exported: {Exported}\n")

print(Summary)
REPORT = open(InputFile + '.summary.txt', 'w')
REPORT.write(Summary)
REPORT.close()
````

## Optional extension — Multiple files supplied from Terminal

Modify the script so that you can supply any number of FASTA filenames as command-line arguments. For each file, produce a separate restriction-site and SSR summary, and write SSR-bearing sequences to a FASTA output named after the source dataset. Use `sys.argv[1:]`.

The code below is a **complete standalone script**, shown here as a string so that you can write it to a `.py` file from Jupyter. It uses the same simple file-handling style as the preceding examples.

````python
Script = 'import sys\nimport re\n\nfor InputFile in sys.argv[1:]:\n    IN = open(InputFile, \'r\')\n    OUT = open(InputFile + \'.SSRs.fa\', \'w\')\n    SeqCount = 0\n    EcoRI = 0\n    MseI = 0\n    HindIII = 0\n    ATtotal = 0\n    ATCtotal = 0\n    GAtotal = 0\n    Exported = 0\n\n    for Line in IN:\n        Fasta = Line.strip("\\n")\n        Seq = IN.readline().strip("\\n")\n        SeqCount += 1\n        EcoRI += len(re.findall(\'GAATTC\', Seq))\n        MseI += len(re.findall(\'TTAA\', Seq))\n        HindIII += len(re.findall(\'AAGCTT\', Seq))\n        ATtotal += len(re.findall(r\'(?:AT){4,}\', Seq))\n        ATCtotal += len(re.findall(r\'(?:ATC){4,}\', Seq))\n        GAtotal += len(re.findall(r\'(?:GA){4,}\', Seq))\n        if (re.search(r\'(?:AT){4,}\', Seq) or\n            re.search(r\'(?:ATC){4,}\', Seq) or\n            re.search(r\'(?:GA){4,}\', Seq)):\n            OUT.write(Fasta + "\\n")\n            OUT.write(Seq + "\\n")\n            Exported += 1\n\n    IN.close()\n    OUT.close()\n    Summary = (f"Input file: {InputFile}\\n"\n               f"Number of sequences: {SeqCount}\\n"\n               f"EcoRI: {EcoRI}\\nMseI: {MseI}\\nHindIII: {HindIII}\\n"\n               f"AT SSRs: {ATtotal}\\nATC SSRs: {ATCtotal}\\nGA SSRs: {GAtotal}\\n"\n               f"Sequences exported: {Exported}\\n")\n    print(Summary)\n    REPORT = open(InputFile + \'.summary.txt\', \'w\')\n    REPORT.write(Summary)\n    REPORT.close()\n'
OUT = open('cutsite_micro_multi_exportSSR.py', 'w')
OUT.write(Script)
OUT.close()
print('Wrote cutsite_micro_multi_exportSSR.py')
````

Run the script in Terminal from the directory containing the FASTA files:

```bash
python cutsite_micro_multi_exportSSR.py Manacus_vitellinus.gene.cds.fa taeda_unigenes.fasta
```

**Notes for discussion:** `re.findall()` counts nonoverlapping occurrences; the regular expressions here match only the sequences as written (not reverse complements). The paired-line approach is intended for the course files **if they have one sequence line per header**. For wrapped/multiline FASTA records, a different parser is needed.
