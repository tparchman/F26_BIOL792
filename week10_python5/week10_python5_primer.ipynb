## Using `sys.argv` to process multiple files.

**This is the best way to process many files at a time when writing and storing code in python scripts (not `jupyter`)**

Remember, `sys.argv` is simply a list of strings provided on the command line when a python script is executed. For the example below, sys.argv[0] would be `file_proc.py`, `sys.argv[1]` would be `file1.txt`, `sys.argv[2]` would be `file2.txt`, and `sys.argv[3]` would be `file3.txt`. 

    $ python file_proc.py file1.txt file2.txt file2.txt

So, opening file connections to each of those files simply requires a `for` loop to work through `sys.argv[1:]`, where `[1:]` means from the `1` element to the end of the list. Anytime you want to do this, something similar to below will do the trick.

```py
    for filename in sys.argv[1:]:
        IN = open(filename, 'r')
```

To build on the above example, lets look at some code which illustrates how to open file connections in a `for` loop using `sys.argv`, and then looping through each file, one line at a time. I have added the print statement below in order to print each scalar (in this case, the file names provided as command line arguments) in the `sys.argv` list as the for loop iterates.

```py
    import sys
    for file in sys.argv[1:]:
	    print(file)
	    IN=open(file, 'r')
	    for Line in IN:
		    Line=Line.strip("\n")
		    print(Line)
	    IN.close()
```
## Adding statement to open and write to new files *each time through* the loop

The code below is similar to above, but statements have been added to open a file to write to **each** time through the loop. This will create a new output file for each separate input file. So, if you processed 100 input files, you would create 100 output files. 

```py
    import sys
    for file in sys.argv[1:]:
        print(file)
        IN=open(file, 'r')
        OUT=open("out_" + file, 'w')
        for Line in IN:
            Line=Line.strip("\n")
            OUT.write(Line + "\n")
        IN.close()
        OUT.close()

```

Notice above in the statment `OUT=open("out_" + file, 'w')` is creating file names that have `out_` appended to the file name variables contained in `sys.argv`. This means that output files will match input file in filenames, albeit with `out` added as a prefix. Thus, if the code below was run, the files `out_file1.txt`, `out_file3.txt`, and `out_file3.txt` would be generated.

    $ python file_proc.py file1.txt file2.txt file2.txt

## Reading and writing safely with the `codecs` module

When working with text files that come from **different operating systems**, **older programs**, or **international datasets**, it’s common to encounter problems with **character encodings** — for example, when a file contains accented letters, Unicode symbols, or non-UTF-8 characters.  

The `codecs` module provides a safe, consistent interface for reading and writing files with specific encodings, such as UTF-8, Latin-1, or ASCII. This helps prevent errors like `UnicodeDecodeError` when Python tries to open a file using the wrong default encoding.

   The `codecs` library facilitates some safeguards on reading from files that may have multiple types of text embedded (ASCII, utf-8). For example, ASCII is often interpretted as utf-8, but the converse is not true. Using this library (or `io.open`) is a safer way to read in files that may have some strangeness. As time goes on, you will likely learn more on when you want to use alternative `open` functions, but for now just trust me. For this weeks data manipulation problem, you can save yourself a headache by using the code below to open file handles while looping through `sys.argv` or a filelist generated with `glob`.

```py
    import codecs
    import sys
    OUT=open("out.txt", 'w')
    for file in sys.argv[1:]:
	    IN = codecs.open(file, 'r', encoding='utf-8', errors='ignore')
	    for Line in IN:
		Line=Line.strip("\n")
		OUT.write(Line + "\n")
	    IN.close()
    OUT.close() 
```



## Printing feedback using `sys.stderr.write()`

Print statements are sometimes used to print messages to screen while a program is running, but they are also often used to print data meant be redirected to a file (using `>` or `>>`). `sys.stderr.write()` allows for printing to screen through a different stream, which wont interfere with output being generated with `print`. This is a useful function for sending updates to screen while your program is running, and is especially useful as a debugging tool. Usage is similar to `.write()`, as demonstrated below.

```py
    import sys
    for file in sys.argv[1:]:
	    sys.stderr.write("Processing file: %s \n" %(file))
	    IN=open(file, 'r')
	    OUT=open("out_"+ file, 'w')
	    sys.stderr.write("Opened file for writing: out_%s \n" %(file))
```

````python
import sys
Nums = range(0,100,5)
for file in Nums:
    sys.stderr.write(f"Processing num: {file} \n" )
````

## Using the `glob` library to find and list many files from jupyter-notebooks

When working with many data files — especially in directories full of `.fasta`, `.txt`, `.csv`, or `.gz` files — the `glob` library provides a simple way to search for files by pattern (using wildcards like `*` and `?`). It is part of Python’s standard library, so you don’t need to install anything, just need to use `import` to call it into use.

Unlike `sys.argv`, which requires you to pass filenames as command-line arguments, `glob` lets you find and process files directly from inside a script or notebook, making it more flexible for exploratory or interactive work.



````python
import glob

files = sorted(glob.glob("*.txt"))
print(files)
print(len(files), "text files found.")
````

### Comparing `sys.argv` and `glob`

Both approaches give us a list of filenames to loop through. The main difference is how we get that list. With `sys.argv`, we supply filenames when running a Python script from the terminal. With `glob`, Python finds filenames matching a pattern, which is convenient in Jupyter notebooks.

From the terminal, a script might be run as `python3 process_files.py file1.txt file2.txt file3.txt`. The script could contain:

```python
import sys

for filename in sys.argv[1:]:
    print(filename)
```

In a notebook, we could instead use:

```python
import glob

files = sorted(glob.glob("*.txt"))
for filename in files:
    print(filename)
```

`sorted()` puts the filenames in a predictable order. This is useful when you want your output to be consistent between runs. Be careful that your `glob` pattern matches the files you actually want to process.


### Finding files in subdirectories

You can include a directory in the pattern, for example `glob.glob("logfiles/*.txt.txt")`. To search through multiple levels of subdirectories, use `**` with `recursive=True`. The pattern must match the actual file extensions: `.txt.txt` is different from `.txt.gz`.


````python
files = sorted(glob.glob("logfiles/*.txt.txt"))
print(len(files), "text files found.")
print(files[:5])  # Look at the first few filenames
````

## Why `glob` is great in Jupyter notebooks

- You don’t need to rerun your notebook with new command-line arguments — just rerun the cell after adding new files to your directory.

- It works naturally with tab completion and dynamic inspection (print(glob.glob("*"))).

- You can easily verify which files were found before opening them, avoiding errors.

- It’s an excellent interactive alternative to sys.argv for early exploratory data analysis.

## Nested loops: working through files and lines

This is the basic structure for processing many files. The **outer loop** goes through filenames. The **inner loop** reads lines within the current file. We open a new input file each time through the outer loop, and close it when we finish reading that file.

The example below uses FASTA files, but the same structure works with many other types of text data. If there are no `.fasta` files in your current directory, the loop will simply have no files to process.


````python
import glob

files = sorted(glob.glob("*.fasta"))

for filename in files:
    IN = open(filename, "r")

    for line in IN:
        line = line.strip("\n")
        print(filename, line)

    IN.close()
````

## Building lists while reading data

We have used lists and `.append()` before. One useful way to organize information is to start with an empty list and add values as we encounter them. Lists that need to accumulate values through a loop must be created **before** that loop.


````python
temperatures = []

for temp in [12.4, 13.5, 14.2, 15.1]:
    temperatures.append(temp)

print(temperatures)
````

We can also build more than one list while reading tabular data. Below, each string represents a row with a sample name and a measurement, separated by a comma. `split()` separates the fields; `.append()` stores them in separate lists.


````python
data_lines = ["sample_A,12.4", "sample_B,13.5", "sample_C,14.2"]

sample_names = []
measurements = []

for line in data_lines:
    fields = line.split(",")
    sample_names.append(fields[0])
    measurements.append(fields[1])

print(sample_names)
print(measurements)
````

## Putting strings together with `join()`

We have used `split()` to break strings into lists. `join()` does the opposite: it joins a list of **strings** into one string, with whatever separator we choose. This is useful when writing comma-separated data.


````python
data = ["H1", "D", "13", "12.4"]
output = ",".join(data)
print(output)
````

If the list contains numbers, convert them to strings first. Here is one way using a list comprehension, which we have already seen:


````python
temperatures = [12.4, 13.5, 14.2]
text_values = [str(temp) for temp in temperatures]
print(",".join(text_values))
````

## Reading many files and writing to **one** output file

Pay attention to where we open and close each file. We open the output file **before** the outer loop so we can keep writing to the same file. We open and close each input file **inside** the outer loop. In this example we write each input line preceded by its filename.

The example uses `.fasta` files so it does not duplicate the specific data manipulation required for this week's assignment.


````python
import glob

files = sorted(glob.glob("*.fasta"))
OUT = open("combined_example.txt", "w")

for filename in files:
    IN = open(filename, "r")

    for line in IN:
        line = line.strip("\n")
        OUT.write(filename + "," + line + "\n")

    IN.close()

OUT.close()
print("Processed", len(files), "files")
````

## Checking our work

A program can run without errors and still produce the wrong output. It is good practice to check how many files were found, inspect the first few lines of output, and compare the number of output lines with what you expected. From a Linux or macOS terminal, `wc -l combined_example.txt` counts output lines. In Python, you can also count them:


````python
IN = open("combined_example.txt", "r")
line_count = 0
for line in IN:
    line_count += 1
IN.close()
print("Output lines:", line_count)
````

## A note on problematic characters in real data

Not all text files are perfectly clean. Some files contain characters that make them difficult to read. You can specify an encoding and how to handle decoding errors when opening a file:

```python
IN = open(filename, "r", encoding="utf-8", errors="ignore")
```

`errors="ignore"` skips characters that cannot be decoded, so use it when necessary and be aware that some information could be lost. A separate problem is an embedded null character (`"\0"`), which can be removed from a line with:

```python
line = line.replace("\0", "")
```

These are two different issues: one concerns decoding the file, and the other concerns cleaning the text after it has been read.


## Extracting information from filenames

Filenames often contain useful information. We can use string methods or regular expressions to extract that information. Here is an example unrelated to the logger files used in this week's assignment:


````python
import re

filename = "sample_17_siteB.fasta"
match = re.search(r"sample_(\d+)_(\w+)\.fasta$", filename)

if match:
    sample_number = match.group(1)
    site = match.group(2)
    print(sample_number, site)

print(filename.replace(".fasta", ""))
````

## Commonly Used Regular expression symbols and metacharacters

| **Symbol / expression** | **Meaning** | **Example** | **Matches** |
|--------------------------|-------------|--------------|--------------|
| `.` | Any single character (except newline) | `r"A.C"` | `"ATC"`, `"AAC"`, `"ACC"` |
| `^` | Start of string or line | `r"^ATG"` | `"ATG..."` only if it appears at start |
| `$` | End of string or line | `r"TAA$"` | Matches `"TAA"` only if it ends the line |
| `[ ]` | Character class | `r"[ATGC]"` | Any one of `A`, `T`, `G`, or `C` |
| `[^ ]` | Negated class | `r"[^0-9]"` | Any character **not** a digit |
| `*` | 0 or more repetitions | `r"A*"` | `""`, `"A"`, `"AA"`, `"AAA"` |
| `+` | 1 or more repetitions | `r"A+"` | `"A"`, `"AA"`, `"AAA"` |
| `?` | 0 or 1 repetition | `r"AT?G"` | `"AG"` or `"ATG"` |
| `{m}` | Exactly *m* repetitions | `r"A{3}"` | `"AAA"` |
| `{m,n}` | Between *m* and *n* repetitions | `r"A{2,4}"` | `"AA"`, `"AAA"`, `"AAAA"` |
| `|` | Logical OR | `r"A|T"` | Matches `"A"` or `"T"` |
| `( )` | Grouping / capture | `r"(ATG)+"` | One or more `"ATG"` repeats |
| `\d` | Any digit | `r"\d+"` | `"123"`, `"007"` |
| `\D` | Any non-digit | `r"\D"` | `"A"`, `"_"` |
| `\w` | Any alphanumeric character | `r"\w+"` | `"A1B2"`, `"gene123"` |
| `\W` | Any non-alphanumeric | `r"\W+"` | `"@"`, `"#"`, `"%"` |
| `\s` | Whitespace (space, tab, newline) | `r"\s+"` | `" "`, `"\t"`, `"\n"` |
| `\S` | Non-whitespace | `r"\S+"` | `"ATGC"`, `"gene"` |
