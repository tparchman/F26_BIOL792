## II. From Python lists to NumPy arrays

We have spent several weeks manipulating lists with loops. NumPy arrays let us perform mathematical operations on many values at once (**vectorization**). NumPy arrays generally contain values of a common data type and can have multiple dimensions.

````python
temperatures = [12.5, 14.2, 16.1, 18.4]
new_temps = []
for temp in temperatures:
    new_temps.append(temp + 2)
print("Using a list:", new_temps)

temperatures_np = np.array(temperatures)
print("Using an array:", temperatures_np + 2)
````

### ndarrays

#### Create 1-D array with `.array()` function
Array is kind of just another word for list. As in base python, the array is given as a list of values enclosed in `[]`.

`type` returns the type of numpy object. IN the case below, an ndarray.

````python
d1 = np.array([1,3,5,2,4,6])
print(d1)
type(d1)
````

#### Create 2-D array with `.array()` function

This is essentially a list of lists, essentially a 2-dimensional data frame of rows with their values being in columns.

````python
d2 = np.array([[1,3,5],[2,4,6]])
print(d2)
type(d2)
````

#### Get dimensions and total size of any `ndarray`

````python
print(d2.shape) #rows, columns
print(d2.size) 
d2.shape
````

### Accessing and indexing `numpy.ndarrays` is similar to lists but with added dimension. Very similar to R indexing of data frames.

This means that `0` is the first index in an array. For a 2-dimensional `numpy.ndarray`, rows are given with the first index, and columns are given with the second: `array_name[row,column]`. This is so similar to dataframes in R, that if you know one the other is quite easy.

````python
print(d2)

# 2nd row, 1st column 
print(d2[1,0]) 

# 1st row, 3rd column
print(d2[0,2])
````

#### Slicing works very similarly to lists in base python. 

This is similar to extracting part of a single row, or collection of rows, or part of a single column, or collection of columns.


````python
#extract first 2 elements of the 2nd row
print(d2[1,:2])
print(d2)
````

### A few properties worth knowing

- `.shape`: size of each dimension
- `.ndim`: number of dimensions
- `.size`: total number of elements
- `.dtype`: data type stored in the array

````python
print("shape:", d2.shape)
print("dimensions:", d2.ndim)
print("elements:", d2.size)
print("dtype:", d2.dtype)

# For a two-dimensional array, the first index selects rows and the second columns.
print("first column:", d2[:, 0])
print("second row:", d2[1, :])
````

### Reshape array format with `.reshape()`

Reshaping an array involves changing its dimensions. For example, we might want to turn a long 1-d array into a 2-d array. 

The code below takes a 1-d array and turns into a 2-d array with 2 rows, and 3 columns.

````python
d1 = np.array([1, 3, 5, 2, 4, 6])
d2_reshaped = d1.reshape(2, 3)
print("original:", d1)
print("reshaped:\n", d2_reshaped)
# reshape returns a reshaped array; assigning it does not change d1.shape
````

The code below takes a 2-d array  with 2 rows, and 3 columns and turns it into a 1-d array.

````python
d2 = np.array([[1,3,5],[2,4,6]])
d2.reshape(1,6)  
````

### create 1D array containing a range of values using the `.arange()` function. 

This function simply returns "A Range" of values.

````python
np.arange(10)
````

#### array 0 to 50 by 5 (start,stop,step)

````python
np.arange(0,50,5)
````

### Boolean indexing: selecting observations that meet a condition

A comparison can return an array of `True` and `False` values. We can use that array to select observations. This will be especially useful when filtering DataFrames next week.

````python
temperatures = np.array([12.5, 14.2, 16.1, 18.4])
warm = temperatures > 15
print(warm)
print(temperatures[warm])
print(temperatures[temperatures > 15])
````

## III. Random numbers and mathematical operations

Random numbers are useful for simulations, sampling, and permutation-based analyses. The modern NumPy interface uses a random-number generator created with `np.random.default_rng()`. Supplying a seed makes examples reproducible.

````python
rng = np.random.default_rng(42)
print(rng.random(5))                    # five floats between 0 and 1
print(rng.integers(0, 100, size=7))     # seven integers
print(rng.integers(0, 100, size=(2, 3)))
print(rng.choice([3, 5, 7, 9], size=4, replace=True))
print(rng.normal(0, 1, size=5))
````

## Math in `numpy`

`numpy` is built for **fast, vectorized math** on arrays (no Python loops). Below are the patterns students will use constantly: element-wise operations, reductions (sum/mean…), broadcasting, and handling NaNs. 

Vectorized math is not just cleaner; it’s much faster than Python for loops.

(https://numpy.org/doc/stable/reference/routines.math.html)

### Generate 100 random numbers from 1 to 1000 and get length, max, min, mean

````python
x = random.randint(1000,size=100)

print(len(x))
print(x.max())
print(x.min())
print(x.mean())
````

### Mathematical operations can easily and rapidly be conducted on each element of an array. 

Below, we are making a 1-d array of ten random integers. We are then using `np.divide` to divide each element by 2, `np.power` to raise each element to the power of 2, and `np.sqrt` to take the square root of each element. 

````python
x = random.randint(1000,size=10)
x2= x.reshape(2,5) #reshape into a 2d array with 2 rows and 5 columns, just for fun.
print(x2)
print(np.divide(x2, 2))

print(np.power(x2, 2))

print(np.sqrt(x2))
````

### Math can be done on multiple arrays

````python
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([10, 20, 30, 40], dtype=float)

x + y          # array([11., 22., 33., 44.])
x * y          # array([ 10.,  40.,  90., 160.])
````

### Reductions (aggregations) and axes

Use reductions to collapse an array along an axis (rows vs columns). This functionality is useful to conduct calculations by row or column, for examples (sums, means, etc.)

````python
rng = np.random.default_rng(42)
A2 = rng.integers(0, 100, size=(4, 5))
print(A2)
print("sum of all:", A2.sum())
print("column sums:", A2.sum(axis=0))
print("row sums:", A2.sum(axis=1))
print("column means:", A2.mean(axis=0))
print("row means:", A2.mean(axis=1))
````

**Remember:** `axis=0` summarizes down rows to produce one value **per column**; `axis=1` summarizes across columns to produce one value **per row**.

### NaN-aware math

 
When missing values appear (NaN), use the `np.nan`* variants of `numpy` math functions

````python
b = np.array([1., 2., np.nan, 4.])
np.mean(b)         # nan  (contaminated)
print("mean:", np.nanmean(b))     
print("SD:", np.nanstd(b))
print("max:", np.nanmax(b))
````

## Iterating through numpy arrays

Just as you can iterate through lists or other data structures in base python, `numpy` ndarrays offer a more rapid environment for iterating. This can mean iterating through a 1-d array or 2-d array, or single 'columns' or 'rows' of 2-d arrays. The idea here is to speed up dataframe operations for large-scale data sets.

**NOTE**: Iteration in `numpy` is mainly for clarity or special cases. For most real data science tasks, use vectorized operations instead of explicit loops — it’s cleaner and much faster.

### Iterating through a 1-d array

````python
arr = np.array([1, 2, 3, 4, 100, 150])

for x in arr:
  print(x) 
````

### Iterating through a 2-d array

````python
arr = np.array([[1, 2, 3], [4, 100, 150]])

for x in arr:
  print(x) 
````

### Iterating through one 'column' of a 2-d array

`arr[0:,2]` below specifies first, all rows, and second, the third column

````python
arr = np.array([[1, 2, 3], [4, 100, 150]])

for x in arr[0:,2]:
  print(x) 
````

## Building `numpy` ndarrays from files, writing ndarrays to files

Using `numpy`, we can quickly read 1-dimensional or 2-dimensional data from text files into ndarray objects. This is very similar to using `read.csv`, `read.delim`, or `read.table` in `R` for those of you with `R` experience.

Lets start by reading a .csv text file into a 2-d array object. We will use a simple file with lat and long for 15 stands of *Pinus muricata* from the coast of California that resides in the same directory as this notebook.

`np.loadtxt` is used to read delimited text that does not have missing values. To control for missing values, `np.genfromtxt` can be used to load data from a text file, with missing values handled as defined. `np.loadtxt` works the same as `np.genfromtxt` when there is no missing data.

````python
x2d = np.loadtxt("muricata_pops_lat_long.csv", delimiter=",")
print(x2d)
````

````python
for i in x2d[0:,0]:
    print("Latitude is: ", i)
````

### Writing ndarray to file:

````python
x = x2d[0:,0] # making 1-d array from twodarray, this will contain just latitudes
print(x)
np.savetxt("np_savetxt_test.txt", x)
````

````python
# Verify that the output file was created (run in Jupyter)
from pathlib import Path
print(Path("np_savetxt_test.txt").exists())
````

### Reading numerical files: practical notes

The preceding *Pinus muricata* examples use `muricata_pops_lat_long.csv`, which should be in the notebook's working directory. `np.loadtxt()` works well for clean numerical tables; `np.genfromtxt()` can handle some missing values. `np.savetxt()` writes arrays to text files. For data containing text columns, pandas will usually be easier (next week).

````python
# Example with missing numerical data (self-contained)
from io import StringIO
example = StringIO("12.5,3.1\nNA,4.2\n15.0,5.4\n")
numeric_data = np.genfromtxt(example, delimiter=",", missing_values="NA", filling_values=np.nan)
print(numeric_data)
print("mean of first column:", np.nanmean(numeric_data[:, 0]))
````

### Preview: combining multiple files with pandas

This connects to last week's work with `glob` and multiple files. We will examine DataFrame manipulation in detail next week. The following cell expects a `sensor_chunks/` directory containing `sensor_chunk_*.csv` files; it does not change your working directory.

````python
from pathlib import Path
files = sorted(glob.glob("sensor_chunks/sensor_chunk_*.csv"))
print("Found", len(files), "files")
if files:
    dfs = [pd.read_csv(filename) for filename in files]
    merged = pd.concat(dfs, ignore_index=True)
    print("Merged shape:", merged.shape)
    print(merged.head())
else:
    print("Optional example: sensor_chunks/ data not found.")
````
