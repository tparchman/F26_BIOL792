## Function for a safe and reusable reader for potentially messy instrumentation data


- Opens each file in binary mode to scrub low-level gremlins (NULL bytes, e.g., `\x00`, `\0`).

- Tries UTF-8 first, then Latin-1 (handles the ° symbol from the header lines in this case).

- Lets pandas auto-detect the delimiter (sep=None, engine="python").


```python
def read_clean_df(path):
    with open(path, "rb") as f:
        raw = f.read().replace(b"\x00", b"")   # remove NULLs
    try:
        text = raw.decode("utf-8")
    except UnicodeDecodeError:
        text = raw.decode("latin-1")
    return pd.read_csv(io.StringIO(text), sep=None, engine="python")
```

#### Read files using `glob`
- Code below will store all filenames in directory ending in `.txt.txt` into `filelist`
- for loop below is just printing elements of the list to confirm this is working as expected.
- Be sure that the current working directory for the jupyter notebook is the directory where the desired `*txt.txt` files reside


```python
filelist = glob.glob('*.txt.txt')
for file in filelist:
    print(file)
```

## Initialize a container for per-file “wide” frames

- Creates an empty list to hold one 2-column frame per file: an index (timestamp/key) + one value column.

- We will later use something like “cbind” (column-bind) these by their index later.

```python
wide_frames = []
```

## Main chunk of code is a loop: read each file, build a key, isolate one value column


- Reads the file with `read_clean_df` function generated above.

- Builds a key (prefer timestamp; fall back to row number if needed).

- Sets the DataFrame’s index to this key.

- Picks a single value column to represent that file (last column by default, adjust if needed).

- Renames that value column to the file’s stem (so columns are unique and traceable).

- Appends the resulting 1-column frame to wide_frames.

- **Indexing by a shared key is the foundation for reliable merges in `pandas`**.

```python
for fn in filelist:
    df = read_clean_df(fn)

    # Build a timestamp key if Date/Time exist (adjust to your columns)
    if {"Date", "Time"}.issubset(df.columns):
        key = pd.to_datetime(df["Date"] + " " + df["Time"], errors="coerce")
    elif "Date Time" in df.columns:
        key = pd.to_datetime(df["Date Time"], errors="coerce")
    else:
        # Fall back to row number if no obvious key (least ideal)
        key = pd.RangeIndex(len(df))

    df_keyed = df.copy()
    df_keyed.index = key

    # Keep a single value column per file (adjust if you want a different one)
    value_col = df_keyed.columns[-1]  # e.g., last column is the reading
    col_name = os.path.splitext(os.path.splitext(os.path.basename(fn))[0])[0]
    df_single = df_keyed[[value_col]].rename(columns={value_col: col_name})

    wide_frames.append(df_single)
```

### To check to see how many 1-column frames were generated, and to see what they look like, try:

```python
# How many frames did we create?
len(wide_frames)

# Peek at one of them
wide_frames[0].head()
```

## Column-bind (outer join) all per-file frames on the key

- `pd.concat(wide_frames, axis=1)` merges by index (the timestamp/key).

- outer join (the default) keeps all timestamps that appear in any file.

- This matters because real sensors don’t always sample at identical times; an outer join preserves everything. This is also why there may be many empty spaces in the data frame.

```python
merged = pd.concat(wide_frames, axis=1)
merged.shape
```

## Optional sort by time:

```python
# Optional: sort by time if the index is a datetime
if pd.api.types.is_datetime64_any_dtype(merged.index):
    merged = merged.sort_index()
```

## writing to a .csv file

- using pandas `pd.to_csv()` function to write the dataframe to .csv file
- print statement followed by the dimensions of the dataframe, using `pd.shape`

- keeping the index means you keep the timestamp (or row key).

- This is a file that can be analyzed in later steps (pandas, plotting, stats).

```python
merged.to_csv("cbind_by_key.csv", index=True)  # keep index (timestamp) in the file
print("Wrote cbind_by_key.csv with shape:", merged.shape)
```

# Keep for now

```python
import glob, os, io
import pandas as pd

def read_clean_df(path):
    with open(path, "rb") as f:
        raw = f.read().replace(b"\x00", b"")   # remove NULLs
    try:
        text = raw.decode("utf-8")
    except UnicodeDecodeError:
        text = raw.decode("latin-1")
    return pd.read_csv(io.StringIO(text), sep=None, engine="python")

filelist = glob.glob("*.txt.txt")
print("Found", len(filelist), "files")

wide_frames = []
for fn in filelist:
    df = read_clean_df(fn)

    # Build a timestamp key if Date/Time exist (adjust to your columns)
    if {"Date", "Time"}.issubset(df.columns):
        key = pd.to_datetime(df["Date"] + " " + df["Time"], errors="coerce")
    elif "Date Time" in df.columns:
        key = pd.to_datetime(df["Date Time"], errors="coerce")
    else:
        # Fall back to row number if no obvious key (least ideal)
        key = pd.RangeIndex(len(df))

    df_keyed = df.copy()
    df_keyed.index = key

    # Keep a single value column per file (adjust if you want a different one)
    value_col = df_keyed.columns[-1]  # e.g., last column is the reading
    col_name = os.path.splitext(os.path.splitext(os.path.basename(fn))[0])[0]
    df_single = df_keyed[[value_col]].rename(columns={value_col: col_name})

    wide_frames.append(df_single)

# Outer-join on the key (keeps all timestamps that appear in any file)
merged = pd.concat(wide_frames, axis=1)

# Optional: sort by time if the index is a datetime
if pd.api.types.is_datetime64_any_dtype(merged.index):
    merged = merged.sort_index()

merged.to_csv("cbind_by_key.csv", index=True)  # keep index (timestamp) in the file
print("Wrote cbind_by_key.csv with shape:", merged.shape)
```
