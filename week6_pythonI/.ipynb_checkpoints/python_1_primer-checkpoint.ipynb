### Splitting strings into pieces with `split()`

Another useful string method is `split()`. The `split()` method splits a string into smaller pieces at a specified separator and returns the pieces as a list.

This is common when working with biological data because data are often stored as lines of text containing multiple values separated by commas, tabs, spaces, or other characters.

For example, imagine that one line of a data file contains:

````python
data = "wolf,12,24.5,Nevada"
````

We can split this string wherever a comma occurs:

````python
fields = data.split(",")
print(fields)
````

The original value was **one string**:

````python
"wolf,12,24.5,Nevada"
````

After using `split(",")`, we have a list containing four separate strings. The character used to divide the string is often called a delimiter. If no separator is specified, `split()` splits a string at whitespace:

````python
data = "wolf 12 24.5 Nevada"
fields = data.split()
print(fields)
````

### Removing characters from the beginning or end of a string with `strip()`

`str.strip()` removes unwanted characters from the **beginning and end** of a string. By default, it removes whitespace characters, including spaces, tabs (`\t`), and newline characters (`\n`). This becomes especially useful when we begin reading data from files, because lines read from a text file will often contain a `\n` newline character at the end.

In the example below, `\n` represents a newline character. The `strip()` method removes it from the end of the sequence.

````python
seq = "ATGCGTA\n"
seq = seq.strip()

print(seq)
````

### Why is `split()` useful?

Breaking lines of text into individual values is an extremely common step when working with data. Once the values have been placed into a list, Python can access and process each value separately using a `for` loop, which is what we start to get into next week.

````python
for value in fields:
    print(value)
````

## 4B. f-strings: formatting variables in print statements

**f-strings** provide a simple way to combine text and variables in a `print()` statement. Place an `f` immediately before the opening quotation mark, then place variable names inside curly brackets `{}` wherever you want their values to appear.

For example, `f"The value is {x}"` will insert the current value of `x` into the string. You can include strings, integers, floats, or even simple calculations inside `{}`. For floats, you can also control the number of digits printed after the decimal using syntax such as `{x:.2f}` for two decimal places.

Inserting a string into printed output:

````python
name = "Lebron"
print(f"Player name: {name}")
````

Instering several variables of different types:

````python
name = "Lebron"
points = 38
assists = 10
rebounds = 16

print(f"{name}: {points} points, {assists} assists, {rebounds} rebounds")
````

Importantly, you can control the number of decimal places for irritatingly long floats:

````python
value = 2 / 3

print(f"2/3 with full precision: {value}")
print(f"2/3 rounded to three decimal places: {value:.3f}")
````

Combining strings, integers and floats:

````python
sequence = "ATGCGCGTA"
length = len(sequence)
gc_content = (sequence.count("G") + sequence.count("C")) / length

print(f"Sequence: {sequence}")
print(f"Length: {length} bases")
print(f"GC content: {gc_content:.3f}")
````

# 7. Examples working with strings in some bioinformatic contexts

Below are short examples that use hard-coded DNA strings. They are meant to reinforce string methods without changing how this notebook is organized.


**DNA as text: length, indexing, slicing**

````python
dna = "ACGTACGTGACGTTTACGTA"
print(f"Sequence: {dna}")
print(f"Length:, {len(dna)}")
print(f"First base: {dna[0]}")
print(f"Last base: {dna[-1]}")
print(f"Bases 5..10 (0-based): {dna[5:10]}")
print(f"Every 3rd base (frame demo): {dna[0::3]}")
````

**Cleaning and case**

````python
raw = "  acgtACGT  \nAC gt  "
clean = raw.upper().replace(" ", "").strip()
print("Raw:", repr(raw))
print("Clean:", clean)
````

**Base counts and GC% (float precision)**

````python
seq = "ACGTACGTGACGTTTACGTA".upper()
a, c, g, t = seq.count("A"), seq.count("C"), seq.count("G"), seq.count("T")
gc = g + c
gc_pct = 100.0 * gc / len(seq)
print(f"A:{a} C:{c} G:{g} T:{t} | length:{len(seq)}")
print(f"GC count = {gc}, GC% = {gc_pct:.2f}%")
````

**Reverse complement with translate**

````python
seq = "ACGTACGTGACGTTTACGTA".upper()
comp_map = str.maketrans({"A":"T", "C":"G", "G":"C", "T":"A"})
complement = seq.translate(comp_map)
rev_comp = complement[::-1]
print("Seq:        ", seq)
print("Complement: ", complement)
print("Rev. comp:  ", rev_comp)
````

**Transcription (DNA→RNA) and a quick codon frame demo**

````python
dna = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"
rna = dna.replace("T", "U")
frame0 = [dna[i:i+3] for i in range(0, len(dna), 3)]
print("DNA:", dna)
print("RNA:", rna)
print("Frame 0 codons (first 10):", frame0[:10])
````

**FASTA-like header and sequence parsing (string only)**

````python
fasta_text = ">seq1 | Columba livia\nACGTACGT\nGACGTTTA\nCGTA\n"
lines = [ln.strip() for ln in fasta_text.splitlines() if ln.strip()]
header = lines[0]
sequence = "".join(ln for ln in lines[1:] if not ln.startswith(">")).upper()
print("Header:", header)
print("Sequence:", sequence, "| length:", len(sequence))
````

**Float formatting patterns**

````python
pi = 3.141592653589793
p = 0.0374231
print(f"pi default: {pi}")
print(f"pi (2 d.p.): {pi:.2f}")
print(f"pi (4 d.p.): {pi:.4f}")
print(f"p as percent (2 d.p.): {p:.2%}")
vals = [1.2, 10.2345, 1000.0/3]
for v in vals:
    print(f"{v:8.2f}")  # width=8, 2 decimals
````

**`help()` and `dir()` (targeted)**

````python
print("Some string methods (first 20 that are public):")
public = [m for m in dir(str) if not m.startswith("_")]
print(public[:20])
print("\nHelp on str.replace:")
help(str.replace)
````

**Optional: tiny k-mer preview (sliding window)**

````python
seq = "ACGTACGTGACGTTTACGTA".upper()
k = 3
kmers = [seq[i:i+k] for i in range(len(seq) - k + 1)]
print("First 10 k-mers:", kmers[:10])
print("Total k-mers:", len(kmers))
````

## 5. Interactively testing and exploring Python code

So far, we have mostly discussed Python code as part of a script (`.py` file). When developing code, however, it is often useful to run individual statements and immediately see what they do. There are two convenient ways to do this: the **Python interactive prompt** in the terminal and **Jupyter notebooks**.

### The Python interactive prompt

From a terminal, typing `python3` starts an interactive Python session:

```bash
$ python3
```

You will see a prompt that looks something like:

```text
>>>
```

You can now enter Python statements one at a time. Python evaluates each statement immediately, making the interactive prompt useful for **quick tests, checking syntax, experimenting with functions or methods, and figuring out why a piece of code is not behaving as expected**.

For example:

```python
>>> seq = "ATGCGCGTA"
>>> len(seq)
9
>>> seq.count("G")
3
>>> seq.lower()
'atgcgcgta'
```

This is a useful way to answer a quick question while you are working at the command line. You do not need to create a script just to test one or two lines of Python.

To leave the interactive Python session, use:

```python
>>> exit()
```

## Jupyter notebooks: interactive Python with a permanent record

A Jupyter notebook provides the same basic ability to run Python interactively, but organizes your work into **cells**. Code in a cell can be run immediately, and the output appears directly below it.

For example, put the following in a Jupyter code cell and run it:

````python
seq = "ATGCGCGTA"
len(seq)
````

Unlike the simple terminal prompt, however, a Jupyter notebook allows you to **save your code, output, explanatory text, equations, figures, and notes together in a single document**. This makes notebooks particularly useful for exploring data, developing analyses, documenting what you did, and sharing reproducible work.

You can also change a cell and run it again:

````python
seq = "ATGCGCGTA"

g = seq.count("G")
c = seq.count("C")
gc_content = (g + c) / len(seq)

print(f"GC content = {gc_content:.3f}")
````

## An important feature of interactive Python

Variables that you create remain available during your interactive session. For example, after running the cell above, you can create a new cell containing only:

````python
gc_content
````

Python already knows what `gc_content` is because it was defined in a cell that you previously ran.

Try another cell, and also note that `tab` autocomplete is useful in this context:

````python
gc_content * 100
````

This ability to build and test an analysis a few lines at a time is one of the major advantages of Jupyter.

**However, remember that the order in which cells are run matters.** If you restart the Jupyter kernel, the variables stored in memory disappear. A well-organized notebook should therefore work correctly if you start at the top and run the cells in order.

## Which should you use?

For this course, we will use all three approaches:

- **Python scripts (`.py`)** when we want to write and save a program that can be run from the command line.
- **The Python interactive prompt** for quick experiments and tests while working in the terminal.
- **Jupyter notebooks** when we want to interactively develop code while keeping our code, results, and explanations together.

Learning to move comfortably among these environments is useful. The Python language is the same in each, the difference is how you want to interact with and organize the code.

## Exploring Python: `help()`, `dir()`, and tab completion

You will **not** memorize every Python function or every method available for different types of objects. An important part of learning to program is learning how to ask Python what tools are available and how they work.

Python provides several ways to explore objects and get help while you are working. Three particularly useful approaches are:

- `type()` — tells you what type of object you are working with.
- `dir()` — shows the attributes and methods available for an object.
- `help()` — provides documentation describing how an object, function, or method works.

In Jupyter, **tab completion** provides an especially convenient way to explore the methods available for an object.

Have a look at the sequence below. Because `seq` is a string, Python knows that all of the built-in string methods are available for it. We can use `dir()` to ask Python what operations are associated with this object:

````python
seq = "ATGCGCGTA"
type(seq)
````

````python
dir(seq)
````

Have a look at what `dir` is doing above.

The resulting list is long, and you do not need to understand everything in it. For now, notice some methods that we have already used, such as `count`, `lower`, `replace`, `strip`, and `upper`.

If you find a method but don't remember exactly how it works, `help()` can provide more information. For example:

````python
help(seq.count)
````

This displays documentation for the `count()` method, including what arguments it accepts and what it returns.

Once you know how the method works, you can immediately test it:

````python
seq.count("G")
````

## Tab completion in Jupyter

Jupyter provides an even faster way to explore an object. In a code cell, type the name of a variable followed by a period:

```text
seq.
```

Then press **Tab**. Jupyter will display the methods and attributes available for that object. You can begin typing the name of a method to narrow the choices.

For example, after finding `replace()` using tab completion, we could try:

````python
seq.replace("T", "U")
````

You can then use `help()` if we want more information:

````python
help(seq.replace)
````

## Try exploring a different type of object

The methods available to an object depend on its type. Compare our string with a float:

````python
seq = "ATGCGCGTA"
allele_frequency = 0.37
print(type(seq))
print(type(allele_frequency))
````

````python
dir(seq)
````

### So.......
When you don't remember how to do something in Python, explore before assuming you need to look up the answer. Check the object's type, use tab completion or `dir()` to see what is available, and use `help()` to learn how a particular function or method works.

A common workflow is:

What is this?: `type()` 
What can it do?: Tab or `dir()`
How does this method work?: `help()`
Try it.

Developing this habit is much more useful than trying to memorize every Python method you encounter.


## 6. Getting input from the user with `input()`

So far, we have created variables by assigning values directly within our Python code. The `input()` function allows a program to instead ask the user to provide a value while the program is running.

For example:

````python
name = input("Enter your name: ")
print(f"Hello {name}")
````

When Python reaches `input()`, the program pauses and displays the text inside the parentheses. Whatever the user types is then stored in the variable `name`.

## An important feature of `input()`: input is returned as a string

Regardless of what you type, `input()` initially returns the value as a **string**. Let's test this rather than just taking Python's word for it.

**Before running the code below, predict what type you think `age` will be if you enter a number such as `25`.**

````python
age = input("Enter your age: ")
print(age)
print(type(age))
````

Even if you entered `25`, Python reports:
```text
<class 'str'>
```
This is important. The value `"25"` is not the same thing to Python as the integer `25`. One is text and one is a number.

Try the following:

````python
age = input("Enter your age: ")
print(age + 5)
````

See hwat happens?

Python cannot add the integer `5` to a string. If we want to use input in a mathematical calculation, we first need to **convert it to the appropriate numeric type**.

For an integer, we can use `int()`:

````python
age = input("Enter your age: ")
age = int(age)
print(age + 5)
````

You can also perform the conversion in a single statement:

````python
age = int(input("Enter your age: "))
print(f"In five years you will be {age + 5}")
````

| `complex()`      | `complex(2,3)` → `(2+3j)`             | Create a complex number.                                               |
| `isinstance()`   | `isinstance(3.14, float)` → `True`    | Checks if variable is of a given type.                                 |

## Common string functions

| Method           | Example                                | Description                                                             |
|------------------|----------------------------------------|-------------------------------------------------------------------------|
| `upper()`        | `"atgc".upper()` → `"ATGC"`            | Convert to uppercase.                                                   |
| `lower()`        | `"ATGC".lower()` → `"atgc"`            | Convert to lowercase.                                                   |
| `capitalize()`   | `"dna".capitalize()` → `"Dna"`         | Capitalize first character.                                             |
| `title()`        | `"dna sequence".title()` → `"Dna Sequence"` | Capitalize each word.                                               |
| `swapcase()`     | `"AtGc".swapcase()` → `"aTgC"`         | Swap upper/lower case.                                                  |
| `count()`        | `"ATGCATGC".count("A")` → `2`          | Count substring occurrences.                                            |
| `find()`         | `"ATGCATGC".find("GC")` → `2`          | Index of first substring match.                                         |
| `rfind()`        | `"ATGCATGC".rfind("GC")` → `6`         | Index of last substring match.                                          |
| `index()`        | `"ATGC".index("T")` → `1`              | Like `find()`, but error if not found.                                  |
| `replace()`      | `"ATGC".replace("A","X")` → `"XTGC"`   | Replace substrings.                                                     |
| `strip()`        | `"  ATGC  ".strip()` → `"ATGC"`        | Trim whitespace (or given chars).                                       |
| `lstrip()`       | `"  ATGC".lstrip()` → `"ATGC"`         | Trim left side.                                                         |
| `rstrip()`       | `"ATGC  ".rstrip()` → `"ATGC"`         | Trim right side.                                                        |
| `startswith()`   | `"ATGC".startswith("AT")` → `True`     | Test start of string.                                                   |
| `endswith()`     | `"ATGC".endswith("GC")` → `True`       | Test end of string.                                                     |
| `split()`        | `"A,T,G,C".split(",")` → `['A','T','G','C']` | Split string into list.                                           |
| `rsplit()`       | `"a b c d".rsplit(maxsplit=2)` → `['a b','c','d']` | Split from right, limited splits.                                  |
| `partition()`    | `"ATGCAT".partition("GC")` → `('AT', 'GC', 'AT')` | Split once into 3 parts.                                          |
| `join()`         | `"".join(['A','T','G','C'])` → `"ATGC"`| Join list of strings.                                                   |
| `center()`       | `"ATGC".center(10,"-")` → `"---ATGC---"` | Center string with padding.                                         |
| `ljust()`        | `"ATGC".ljust(8,".")` → `"ATGC...."`   | Left-align and pad.                                                     |
| `rjust()`        | `"ATGC".rjust(8,".")` → `"....ATGC"`   | Right-align and pad.                                                    |
| `zfill()`        | `"42".zfill(5)` → `"00042"`            | Pad numbers with leading zeros.                                         |
| `isalnum()`      | `"ATGC123".isalnum()` → `True`         | Checks if alphanumeric.                                                 |
| `isalpha()`      | `"ATGC".isalpha()` → `True`            | Checks if all letters.                                                  |
| `isdigit()`      | `"12345".isdigit()` → `True`           | Checks if all digits.                                                   |
| `islower()`      | `"dna".islower()` → `True`             | Checks if all lowercase.                                                |
| `isupper()`      | `"DNA".isupper()` → `True`             | Checks if all uppercase.                                                |
| `isspace()`      | `"   ".isspace()` → `True`             | Checks if string is only whitespace.                                    |
| `istitle()`      | `"Dna Sequence".istitle()` → `True`    | Checks if string is in title case.                                      |
