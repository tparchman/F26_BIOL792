```python
SeqA = "AAAAAAATTGCTTT"
```

```python
Act = SeqA.count('A')
LenA = len(SeqA)
print("count of A:",Act, "Length of SeqA:", LenA)
Start = SeqA.startswith('AAA')
End = SeqA.endswith('AAA')
print("Does SeqA start with AAA?",Start)
print("Does SeqA end with AAA?",End)
```
