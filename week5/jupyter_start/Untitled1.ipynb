### A simple plot demo

Notebooks really shine when you mix code, text, and visualizations. Try this in a new cell:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y)
plt.title("Simple Plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```
