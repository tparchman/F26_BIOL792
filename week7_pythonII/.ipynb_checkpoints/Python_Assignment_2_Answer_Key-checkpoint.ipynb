# 1. Working with sequence information

Start with the following string:

```python
dna_info = "HOWR_017,Nevada,ATATCTCGCGGGGTTTATATATATTATTTAAA"
```

This single string contains three comma-delimited fields: a sample name, a population, and a DNA sequence.


````python
dna_info = "HOWR_017,Nevada,ATATCTCGCGGGGTTTATAatatatattcgcgcgcg"
````

#### A. Use `.split()` to turn `dna_info` into a list. Print the resulting list.


````python
data_list = dna_info.split(",")
print(data_list)
````

#### B. Using list indices, save the sample name, population, and DNA sequence as three separate variables. Print a sentence identifying the sample and population.

Your output should resemble:

```text
Sample HOWR_017 is from Nevada
```


````python
sample = data_list[0]
population = data_list[1]
dna_seq = data_list[2]

print(f"Sample {sample} is from {population}")
````

#### C. Convert the DNA sequence to uppercase. Print the sequence and its length.


````python
dna_seq = dna_seq.upper()

print(f"Sequence: {dna_seq}")
print(f"Sequence length: {len(dna_seq)}")
````

#### D. Count the number of `A`, `C`, `G`, and `T` bases in the sequence. Use these counts to calculate GC content:

```text
GC content = (number of G bases + number of C bases) / total sequence length
```

Print the GC content to three decimal places.


````python
num_a = dna_seq.count("A")
num_c = dna_seq.count("C")
num_g = dna_seq.count("G")
num_t = dna_seq.count("T")

gc_content = (num_g + num_c) / len(dna_seq)

print(f"A: {num_a}")
print(f"C: {num_c}")
print(f"G: {num_g}")
print(f"T: {num_t}")
print(f"GC content: {gc_content:.3f}")
````

#### E. Use an `if`/`else` statement to classify the sequence as:

- `"GC-rich"` if GC content is greater than or equal to `0.50`
- `"AT-rich"` if GC content is less than `0.50`

Print the classification.


````python
if gc_content >= 0.50:
    print("GC-rich")
else:
    print("AT-rich")
````

#### F. Use slicing to save the first 10 bases and the last 10 bases as two new variables. Print both.


````python
first_ten = dna_seq[:10]
last_ten = dna_seq[-10:]

print(f"First 10 bases: {first_ten}")
print(f"Last 10 bases: {last_ten}")
````

# 2. Filtering a list of measurements

Start with this comma-delimited string of measurements:

```python
measurement_data = "12.4,8.7,15.1,4.3,19.8,10.2,6.6,14.5,21.3,9.8"
```


````python
measurement_data = "12.4,8.7,15.1,4.3,19.8,10.2,6.6,14.5,21.3,9.8"
````

#### A. Use `.split()` to convert `measurement_data` into a list. Print the list and its length.

Notice that the elements are still strings.


````python
measurement_strings = measurement_data.split(",")

print(measurement_strings)
print(f"Number of measurements: {len(measurement_strings)}")
````

#### B. Create an empty list called `measurements`. Use a `for` loop to move through the list from part A, convert each value to a `float`, and `.append()` it to `measurements`.

Print the new list.


````python
measurements = []

for value in measurement_strings:
    measurements.append(float(value))

print(measurements)
````

#### C. Create two empty lists:

```python
large_values = []
small_values = []
```

Loop through `measurements`. If a value is greater than or equal to `10`, append it to `large_values`. Otherwise, append it to `small_values`.

Print both lists and the number of elements in each.


````python
large_values = []
small_values = []

for value in measurements:
    if value >= 10:
        large_values.append(value)
    else:
        small_values.append(value)

print(f"Large values: {large_values}")
print(f"Number of large values: {len(large_values)}")
print(f"Small values: {small_values}")
print(f"Number of small values: {len(small_values)}")
````

#### D. Use a `for` loop to calculate the sum of all values in `measurements`.

Start by creating:

```python
total = 0
```

During each trip through the loop, add the current measurement to `total`.

After the loop, calculate the mean using `total` and `len(measurements)`. Print the total and mean.

**Do not use `sum()` for this problem.** The goal is to practice accumulating a value during a loop.


````python
total = 0

for value in measurements:
    total += value

mean = total / len(measurements)

print(f"Total: {total:.2f}")
print(f"Mean: {mean:.2f}")
````

#### E. Use an `if`/`elif`/`else` statement to classify the mean:

- greater than or equal to `15`: `"high"`
- greater than or equal to `10` but less than `15`: `"moderate"`
- less than `10`: `"low"`

Print the classification.


````python
if mean >= 15:
    classification = "high"
elif mean >= 10:
    classification = "moderate"
else:
    classification = "low"

print(f"Mean classification: {classification}")
````

# 3. Processing samples with loops and conditionals

Below is a list containing simplified sample records. Each record contains a sample name, population, and sequence separated by commas.

```python
samples = [
    "HOWR_01,NV,ATGCGCGTAA",
    "HOWR_02,CA,ATATATTTAA",
    "HOWR_03,NV,GCGCGCCGTA",
    "HOWR_04,OR,AATTGGCCAA",
    "HOWR_05,NV,GGGCCCGCTA"
]
```

For this problem, use **one `for` loop** to process all five sample records.


````python
samples = [
    "HOWR_01,NV,ATGCGCGTAA",
    "HOWR_02,CA,ATATATTTAA",
    "HOWR_03,NV,GCGCGCCGTA",
    "HOWR_04,OR,AATTGGCCAA",
    "HOWR_05,NV,GGGCCCGCTA"
]
````

#### A. Inside the loop, use `.split()` to separate each record into sample name, population, and DNA sequence.

For each sample, print a line containing the sample name, population, and sequence length.

For example:

```text
HOWR_01, NV, sequence length = 10
```


````python
for record in samples:
    fields = record.split(",")

    sample = fields[0]
    population = fields[1]
    dna_seq = fields[2]

    print(f"{sample}, {population}, sequence length = {len(dna_seq)}")
````

#### B. Inside the same loop, calculate the GC content of each sequence.

Print the sample name and GC content to two decimal places.

The code below shows the additional operations that belong inside the loop after the variables from Part A have been created.


````python
for record in samples:
    fields = record.split(",")
    sample = fields[0]
    dna_seq = fields[2]

    num_g = dna_seq.count("G")
    num_c = dna_seq.count("C")
    gc_content = (num_g + num_c) / len(dna_seq)

    print(f"{sample} GC content = {gc_content:.2f}")
````

#### C. Still inside the loop, classify each sequence as:

- `"high GC"` if GC content is greater than or equal to `0.60`
- `"intermediate GC"` if GC content is greater than or equal to `0.40` but less than `0.60`
- `"low GC"` if GC content is less than `0.40`

Print the classification for each sample.


````python
for record in samples:
    fields = record.split(",")
    sample = fields[0]
    dna_seq = fields[2]

    gc_content = (dna_seq.count("G") + dna_seq.count("C")) / len(dna_seq)

    if gc_content >= 0.60:
        classification = "high GC"
    elif gc_content >= 0.40:
        classification = "intermediate GC"
    else:
        classification = "low GC"

    print(f"{sample}: {classification}")
````

#### D. Before the loop, create an empty list called:

```python
nv_samples = []
```

During the loop, if the sample is from `"NV"`, append its sample name to `nv_samples`.

After the loop has finished, print:

1. the complete `nv_samples` list
2. the number of Nevada samples


````python
nv_samples = []

for record in samples:
    fields = record.split(",")
    sample = fields[0]
    population = fields[1]

    if population == "NV":
        nv_samples.append(sample)

print(f"Nevada samples: {nv_samples}")
print(f"Number of Nevada samples: {len(nv_samples)}")
````

#### E. Add a counter that keeps track of the number of sequences classified as `"high GC"`.

Initialize the counter **before** the loop and update it at the appropriate place **inside** the loop.

After the loop, print the number of high-GC samples.


````python
high_gc_count = 0

for record in samples:
    fields = record.split(",")
    dna_seq = fields[2]

    gc_content = (dna_seq.count("G") + dna_seq.count("C")) / len(dna_seq)

    if gc_content >= 0.60:
        high_gc_count += 1

print(f"Number of high-GC samples: {high_gc_count}")
````

### Complete solution for Problem 3

Parts A–E are intended to work together in **one loop**. The complete solution looks like this:


````python
nv_samples = []
high_gc_count = 0

for record in samples:
    fields = record.split(",")

    sample = fields[0]
    population = fields[1]
    dna_seq = fields[2]

    print(f"{sample}, {population}, sequence length = {len(dna_seq)}")

    num_g = dna_seq.count("G")
    num_c = dna_seq.count("C")
    gc_content = (num_g + num_c) / len(dna_seq)

    print(f"{sample} GC content = {gc_content:.2f}")

    if gc_content >= 0.60:
        classification = "high GC"
        high_gc_count += 1
    elif gc_content >= 0.40:
        classification = "intermediate GC"
    else:
        classification = "low GC"

    print(f"{sample}: {classification}")

    if population == "NV":
        nv_samples.append(sample)

print(f"Nevada samples: {nv_samples}")
print(f"Number of Nevada samples: {len(nv_samples)}")
print(f"Number of high-GC samples: {high_gc_count}")
````

# 4. Challenge: summarize samples by population

Use the same `samples` list from Problem 3.

This problem intentionally gives you less step-by-step guidance. Break the task into the same small operations used in the earlier problems:

**loop → split → extract or calculate → test → save/count**


#### A. Create three counters called `num_nv`, `num_ca`, and `num_or`, and initialize each to `0`. Write a `for` loop that processes each record, uses `.split()` to identify its population, and updates the appropriate population counter.

After the loop, print the number of samples from each population.


````python
num_nv = 0
num_ca = 0
num_or = 0

for record in samples:
    fields = record.split(",")
    population = fields[1]

    if population == "NV":
        num_nv += 1
    elif population == "CA":
        num_ca += 1
    elif population == "OR":
        num_or += 1

print(f"Nevada samples: {num_nv}")
print(f"California samples: {num_ca}")
print(f"Oregon samples: {num_or}")
````

#### B. Create an empty list called `long_sample_names`. During the same loop (or in a new loop), append the sample name to this list if its name contains more than 6 characters.

After the loop, print `long_sample_names` and the number of sample names it contains.


````python
long_sample_names = []

for record in samples:
    fields = record.split(",")
    sample = fields[0]

    if len(sample) > 6:
        long_sample_names.append(sample)

print(f"Long sample names: {long_sample_names}")
print(f"Number of long sample names: {len(long_sample_names)}")
````

#### C. Create an empty list called `gc_values`. Loop through the sample records, calculate the GC content of each DNA sequence, and append each GC-content value to `gc_values`.

After the loop, print the complete list of GC-content values.


````python
gc_values = []

for record in samples:
    fields = record.split(",")
    dna_seq = fields[2]

    num_g = dna_seq.count("G")
    num_c = dna_seq.count("C")
    gc_content = (num_g + num_c) / len(dna_seq)

    gc_values.append(gc_content)

print(f"GC values: {gc_values}")
````

#### D. Use another `for` loop and a counter to determine how many values in `gc_values` are greater than or equal to `0.50`. Print the final count.


````python
high_gc = 0

for value in gc_values:
    if value >= 0.50:
        high_gc += 1

print(f"Number with GC content >= 0.50: {high_gc}")
````
